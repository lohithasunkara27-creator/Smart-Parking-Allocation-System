```python
import gradio as gr
import heapq
from datetime import datetime

# =====================================================
# PARKING SLOT DATABASE
# =====================================================

parking_slots = {

    "A1": {"type":"Car","status":"Free","distance":2},
    "A2": {"type":"Car","status":"Occupied","distance":4},
    "A3": {"type":"Car","status":"Free","distance":1},

    "B1": {"type":"Bike","status":"Free","distance":3},
    "B2": {"type":"Bike","status":"Occupied","distance":5},

    "C1": {"type":"Truck","status":"Free","distance":2}
}

# =====================================================
# SLOT ALLOCATION
# =====================================================

def allocate_slot(vehicle_no, vehicle_type):

    if vehicle_no.strip() == "":
        return "<h2>Please Enter Vehicle Number</h2>"

    available = []

    for slot in parking_slots:

        if (
            parking_slots[slot]["status"] == "Free"
            and
            parking_slots[slot]["type"] == vehicle_type
        ):

            heapq.heappush(
                available,
                (
                    parking_slots[slot]["distance"],
                    slot
                )
            )

    if len(available) == 0:

        assigned_slot = "No Slot Available"
        status = "Waiting"

    else:

        assigned_slot = heapq.heappop(available)[1]

        parking_slots[assigned_slot]["status"] = "Reserved"

        status = "Reserved"

    current_time = datetime.now().strftime("%d-%m-%Y %H:%M:%S")

    # =====================================================
    # SLOT DISPLAY
    # =====================================================

    slot_html = ""

    for slot in parking_slots:

        slot_status = parking_slots[slot]["status"]

        if slot_status == "Free":
            color = "#16a34a"

        elif slot_status == "Occupied":
            color = "#dc2626"

        else:
            color = "#ca8a04"

        slot_html += f"""

        <div style="
        background:{color};
        padding:20px;
        border-radius:15px;
        text-align:center;
        font-size:20px;
        font-weight:bold;
        color:white;
        ">

        {slot}

        <br><br>

        {slot_status}

        </div>

        """

    html = f"""

    <!DOCTYPE html>

    <html>

    <head>

    <style>

    body {{
        background:#081120;
        color:white;
        font-family:Arial;
        padding:20px;
    }}

    .title {{
        text-align:center;
        font-size:42px;
        font-weight:bold;
        color:#00d9ff;
        margin-bottom:30px;
    }}

    .card {{
        background:linear-gradient(135deg,#1b4332,#2d6a4f);
        padding:30px;
        border-radius:20px;
        margin-bottom:25px;
        box-shadow:0px 0px 15px rgba(0,255,255,0.2);
    }}

    .heading {{
        font-size:30px;
        font-weight:bold;
        margin-bottom:20px;
        color:white;
    }}

    .info {{
        font-size:22px;
        line-height:2;
        color:white;
    }}

    .grid {{
        display:grid;
        grid-template-columns:repeat(3,1fr);
        gap:20px;
        margin-top:20px;
    }}

    .footer {{
        text-align:center;
        font-size:20px;
        margin-top:20px;
        color:#00ffcc;
    }}

    </style>

    </head>

    <body>

    <div class="title">
    🚗 ParkAI — Automated Parking Management System
    </div>

    <div class="card">

    <div class="heading">
    Vehicle Information
    </div>

    <div class="info">

    Vehicle Number :
    <b>{vehicle_no}</b>

    <br>

    Vehicle Type :
    <b>{vehicle_type}</b>

    <br>

    Assigned Slot :
    <b>{assigned_slot}</b>

    <br>

    Parking Status :
    <b>{status}</b>

    <br>

    Entry Time :
    <b>{current_time}</b>

    </div>

    </div>

    <div class="card">

    <div class="heading">
    Parking Slots Status
    </div>

    <div class="grid">

    {slot_html}

    </div>

    </div>

    <div class="footer">

    Smart AI Based Parking Allocation System

    </div>

    </body>

    </html>

    """

    return html

# =====================================================
# GRADIO UI
# =====================================================

demo = gr.Interface(

    fn=allocate_slot,

    inputs=[

        gr.Textbox(
            label="Vehicle Number",
            placeholder="Enter Vehicle Number"
        ),

        gr.Dropdown(
            ["Car","Bike","Truck"],
            label="Vehicle Type"
        )
    ],

    outputs=gr.HTML(),

    title="ParkAI - AI Parking System"
)

demo.launch(share=True)
```
